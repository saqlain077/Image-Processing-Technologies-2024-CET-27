## 1. Import NumPy

```python
import numpy as np
```

## 2. Create a 3 × 3 grayscale image matrix



```python
image = np.array([
    [0, 128, 255],
    [64, 180, 220],
    [255, 100, 0]
])

print("Image matrix:")
print(image)
print("Number of rows:", image.shape[0])
print("Number of columns:", image.shape[1])
```

## 3. Find a pixel value



```python
print("Pixel at row 1, column 2:")
print(image[0, 1])
```

## 4. Calculate total number of pixels

Formula: **Total pixels = Width × Height**

```python
width = 640
height = 480

total_pixels = width * height
print("Total pixels:", total_pixels)
```

## 5. Calculate uncompressed image storage size

For an 8-bit grayscale image: size in bytes = (width × height × bits per pixel) ÷ 8.

```python
width = 640
height = 480
bits_per_pixel = 8

size_in_bits = width * height * bits_per_pixel
size_in_bytes = size_in_bits / 8

print("Size in bits:", size_in_bits)
print("Size in bytes:", int(size_in_bytes))
```

## 6. Calculate RGB image storage size

An RGB image commonly uses 8 bits for each of its three channels (red, green, blue), so it uses 24 bits or 3 bytes per pixel.

```python
width = 3
height = 3
bytes_per_pixel = 3

rgb_size = width * height * bytes_per_pixel
print("Uncompressed RGB pixel-data size:", rgb_size, "bytes")
```

## 7. Increase image brightness

Rule: `g(x, y) = min(f(x, y) + 50, 255)`. NumPy's `clip` keeps values in the valid range 0–255.

```python
image = np.array([
    [0, 100, 200],
    [50, 150, 250]
])

bright_image = np.clip(image + 50, 0, 255)

print("Original image:")
print(image)

print("Brighter image:")
print(bright_image)
```
