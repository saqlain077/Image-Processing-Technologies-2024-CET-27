- **Task II:** Change the text to your own message
- **Task III:** Change image dimensions from (300, 400, 3) to (200, 500, 3)
- **Task IV:** Use red text colour (0, 0, 255) — OpenCV uses BGR order

### Task II + III + IV combined (custom message, 200x500 image, red text)

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

print("OpenCV version:", cv2.__version__)

# Task III: new dimensions (200 height, 500 width, 3 channels)
image = np.zeros((200, 500, 3), dtype=np.uint8)

# Task II: your own message  |  Task IV: red colour (BGR: 0, 0, 255)
cv2.putText(
    image,
    "hello hi tata bye bye",   # <-- change this to your own message
    (30, 100),                      # starting position (x, y)
    cv2.FONT_HERSHEY_SIMPLEX,
    1,                              # font scale
    (0, 0, 255),                    # red in BGR order
    2                               # thickness
)

cv2_imshow(image)
```

## Section 3: Python Basic Image Operations

### Task 1 — Load and display an image (image is a NumPy array)

```python
import cv2, numpy as np, matplotlib.pyplot as plt
from skimage import data

img = data.astronaut()   # RGB demo image

print("Shape (height, width, channels):", img.shape)
print("Data type:", img.dtype)

plt.imshow(img); plt.axis('off'); plt.show()
```

### Task 2 — Display its RGB components

```python
R, G, B = cv2.split(img)

fig, ax = plt.subplots(1, 3, figsize=(10, 3))

ax[0].imshow(R, cmap='Reds'); ax[0].set_title('R')
ax[1].imshow(G, cmap='Greens'); ax[1].set_title('G')
ax[2].imshow(B, cmap='Blues'); ax[2].set_title('B')

for a in ax:
    a.axis('off')

plt.show()
```

### Task 3 — Convert it to grayscale

```python
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

print(gray.shape, gray.dtype)

plt.imshow(gray, cmap='gray'); plt.axis('off'); plt.show()
```

### Task 4 — Apply thresholding (global vs adaptive)

```python
_, binary = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY)

adaptive = cv2.adaptiveThreshold(
    gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY, 31, 5)

fig, ax = plt.subplots(1, 2, figsize=(8, 3))

ax[0].imshow(binary, cmap='gray'); ax[0].set_title('Global')
ax[1].imshow(adaptive, cmap='gray'); ax[1].set_title('Adaptive')

for a in ax:
    a.axis('off')

plt.show()
```
